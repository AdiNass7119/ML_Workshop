```python
#DataVisualization --> matplotlib,seaborn,plotly
import matplotlib.pyplot as plt #!pip install matplotlib
```

```python
#Distribution of Age
plt.hist(data['Age'],bins=15,color='blue')
plt.xlabel("Age")
plt.ylabel("Count")
plt.title("Age Distribution")
plt.show()
```

```python
#Display the count of the bars in histogram
#Distribution of Age
counts,edges,bars = plt.hist(data['Age'],bins=25,color='red')
print(counts)
print(bars)
plt.xlabel("Age")
plt.ylabel("Count")
plt.bar_label(bars)
plt.title("Age Distribution")
plt.show()
```

```python
plt.bar(data['Survived'],data['Sex'])
plt.xlabel("Survived Rate")
plt.ylabel("Gender")
plt.title("Gender Survived Distribution")
plt.show()
```

```python
data['Survived'].value_counts()
```

```python
#549 / (549+342)
index = data['Survived'].value_counts().keys().tolist()
print(index)
labels = data['Survived'].value_counts().values.tolist()
print(labels)
```

```python
#Pie Chart
plt.pie(x = labels,
        labels = ['Dead','Alive'],
       colors = ['red','brown'],
       autopct="%.1f%%",
       shadow=True)
plt.title("Survived Distribution")
plt.legend()
plt.show()
```

```python
#Pie Chart
plt.pie(x = [577,314],
        labels = ['M','F'],
       colors = ['red','blue'],
       autopct="%.1f%%",
       shadow=True)
plt.title("Gender Distribution")
plt.legend()
plt.show()
```

```python
#pairwise relationship --> scatterplot()
plt.scatter(x=data['Age'],
            y=data['Fare'])
plt.xlabel("Age")
plt.ylabel("Fare")
plt.title("Age vs Fare Distribution")
plt.show()
```
